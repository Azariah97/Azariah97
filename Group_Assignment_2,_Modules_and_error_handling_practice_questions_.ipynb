```python
def calculate_car_insurance_premium(age, years_without_accidents, car_value, car_type=None):
    # Step 1: Error handling
    if age < 16 or age > 100:
        raise ValueError("Age must be between 16 and 100.")
    if years_without_accidents < 0 or car_value < 0:
        raise ValueError("Inputs must be non-negative.")

    # Step 2: Initialize base premium
    premium = 500

    # Step 3: Apply age factor
    if age < 25:
        premium += 100
    elif age >= 40:
        premium -= 50

    # Step 4: Apply driving history factor
    for y in range(min(years_without_accidents, 5)):
        premium *= 0.95

    # Step 5: Apply car value factor
    if car_value >= 20000 and car_value < 50000:
        premium *= 1.05
    elif car_value >= 50000:
        premium *= 1.10

    # Step 6 (Bonus): Apply car type factor (if car_type is provided)
    if car_type == "sports car":
        premium *= 1.20  # Example factor for a sports car
    elif car_type == "SUV":
        premium *= 1.10  # Example factor for an SUV
    elif car_type == "sedan":
        premium *= 1.00  # No additional change for sedan

    # Step 7: Round the final premium
    premium = round(premium, 2)

    return premium

# Step 8: Test cases
# Test case 1: Normal case
# Test case 2: Edge case (young driver, expensive car)
# Test case 3: Error case (invalid input)
```
